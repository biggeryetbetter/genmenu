```python
from huggingface_hub import hf_hub_download

hf_hub_download("Ultralytics/YOLO26", "yolo26x.pt", local_dir="./models/")
```

```python
from ultralytics import YOLO

# Load a pretrained YOLO26n model
model = YOLO("models/yolo26x.pt")

# Train the model on the COCO8 dataset for 100 epochs
train_results = model.train(
    data="/home/media/datasets/GenMenuYOLO26/data.yaml",  # Path to dataset configuration file
    epochs=100,  # Number of training epochs
    imgsz=640,  # Image size for training
    device="cpu",  # Device to run on (e.g., 'cpu', 0, [0,1,2,3])
)

# Export the model to ONNX format for deployment
path = model.export()  # Returns the path to the exported model

# Evaluate the model's performance on the validation set
metrics = model.val()

# Perform object detection on an image
results = model("/home/media/datasets/GenMenuYOLO26/images/train/02370.png")  # Predict on an image
results[0].show()  # Display results
```
