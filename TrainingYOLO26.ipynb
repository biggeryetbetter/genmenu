```python
from huggingface_hub import hf_hub_download

hf_hub_download("Ultralytics/YOLO26", "yolo26x.pt", local_dir="./models/")
```

```python
from ultralytics import YOLO

# Load a pretrained YOLO26n model
model = YOLO("models/yolo26x.pt")

# Train the model on the COCO8 dataset for 100 epochs
train_results = model.train(
    data="/home/media/datasets/GenMenuYOLO26/data.yaml",  # Path to dataset configuration file
    epochs=100,  # Number of training epochs
    imgsz=640,  # Image size for training
    device="cpu",  # Device to run on (e.g., 'cpu', 0, [0,1,2,3])
)

# Export the model to ONNX format for deployment
path = model.export()  # Returns the path to the exported model

# Evaluate the model's performance on the validation set
metrics = model.val()

# Perform object detection on an image
results = model("/home/media/datasets/GenMenuYOLO26/images/train/02370.png")  # Predict on an image
results[0].show()  # Display results
```

```python
from ultralytics import YOLO

# Load a pretrained YOLO26n model
model = YOLO("models/best.pt")
```

```python
test_image = "6b603189e6ad76490bb78b4e334ba7b8774b6955.jpg"

# Perform object detection on an image
results = model(test_image)  # Predict on an image
# results[0].show()  # Display results
```

```python
from PIL import Image, ImageDraw

COLORS = [
    (230, 25, 75), (60, 180, 75), (0, 130, 200), (245, 130, 48), (145, 30, 180),
    (70, 240, 240), (240, 50, 230), (128, 128, 0), (0, 128, 128), (255, 215, 0),
]

img = Image.open(test_image).convert("RGBA")
overlay = Image.new("RGBA", img.size, (255, 255, 255, 0))
draw = ImageDraw.Draw(overlay)
for box in results[0].boxes:
    [x1, y1, x2, y2] = list(box.xyxy[0])
    color = COLORS[int(box.cls[0]) % len(COLORS)]
    draw.rectangle((x1, y1, x2, y2), fill=(*color, 40), outline=(*color, 255), width=2)
    draw.text((x1 + 2, y1 + 1), model.names[int(box.cls[0])], fill=(*color, 255))
img.alpha_composite(overlay)
# img.convert("RGB").save(out_path)
img
```
